```python
import matplotlib.pyplot as plt
import numpy as np

# Data for the five folds
folds = ['Fold 1', 'Fold 2', 'Fold 3', 'Fold 4', 'Fold 5']
model1_acc = [0.7196, 0.8157, 0.9006, 0.6554, 0.7885]
model2_acc = [0.7756, 0.6538, 0.8526, 0.7340, 0.7997]
model3_acc = [0.7532, 0.7756, 0.7869, 0.7644, 0.7756]

# Set positions for bars on the x-axis
x = np.arange(len(folds))
width = 0.25

fig, ax = plt.subplots(figsize=(10,6))

# Create the bars for each model
rects1 = ax.bar(x - width, model1_acc, width, label='Model 1')
rects2 = ax.bar(x, model2_acc, width, label='Model 2')
rects3 = ax.bar(x + width, model3_acc, width, label='Model 3')

# Add labels, title, and custom x-axis tick labels
ax.set_ylabel('Test Accuracy')
ax.set_xlabel('Folds')
ax.set_title('Test Accuracies Across 5 Folds')
ax.set_xticks(x)
ax.set_xticklabels(folds)
ax.legend()

# Optionally, add value labels on top of each bar
for rect in rects1 + rects2 + rects3:
    height = rect.get_height()
    ax.annotate(f'{height:.2f}',
                xy=(rect.get_x() + rect.get_width() / 2, height),
                xytext=(0, 3),  # 3 points vertical offset
                textcoords="offset points",
                ha='center', va='bottom')

fig.tight_layout()
plt.show()
```
